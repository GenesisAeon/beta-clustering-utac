# Φ^(1/3) Inter-Cluster Scaling

Tests whether β_cluster(n+1) / β_cluster(n) ≈ Φ^(1/3) ≈ 1.17480


```python
from beta_clustering import BetaClusteringUTAC, PHI_CUBEROOT

system = BetaClusteringUTAC()
system.run_cycle()

result = system._phi_result
print(f'Cluster centre ratios: {[f"{r:.4f}" for r in result.ratios]}')
print(f'Mean ratio:  {result.mean_ratio:.5f}')
print(f'Φ^(1/3):     {PHI_CUBEROOT:.5f}')
print(f'Deviation:   {result.mean_deviation:.5f}')
print(f'p-value:     {result.p_value_approx:.4f}')
print(f'H₀ rejected: {result.null_rejected}')
```
